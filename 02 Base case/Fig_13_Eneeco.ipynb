```python
# Replace with your path if different
df =  pd.read_excel (r'resultsProposed_alpha.xlsx', sheet_name='chBIBminus')

# Initialize nested dictionary
scenario_alpha_data_ch_minus = defaultdict(lambda: defaultdict(list))

#dictionary generation
for i in range(len(df)):
    scenario = df.loc[i, 'scenario']
    alpha = df.loc[i, 'alpha']
    ch_minus = df.loc[i, 'ch_minus']
    scenario_alpha_data_ch_minus[scenario][alpha].append(ch_minus)


# convert to regular dict (for cleaner printing or export)
scenario_alpha_data_ch_minus = {k: dict(v) for k, v in  scenario_alpha_data_ch_minus.items()}

# Print the ch_minus list for scenario 1 and alpha 1
print(scenario_alpha_data_ch_minus[1.0][1.0])
```

```python
X = np.arange(25)


width = 0.9
hw = 0.2
hl = 0.3
arrow_width = 0.05
zero_tolerance = 1e-9

fig, axs = plt.subplots(
    3, 1,
    figsize=(25, 15),
    constrained_layout=True
)


scenarios = [
    (1, -0.33, "deeppink"),
    (5, 0.00, "darkorange"),
    (10, 0.33, "rebeccapurple")
]


def draw_arrow(ax, x, y, dy, color):
    dy = float(dy)

    if not np.isfinite(dy):
        raise ValueError(
            f"Invalid arrow length at x={x}: {dy}"
        )

    if abs(dy) <= zero_tolerance:
        return

    ax.arrow(
        x, y, 0, dy,
        color=color,
        length_includes_head=True,
        head_width=hw,
        head_length=hl,
        width=arrow_width
    )


def draw_panel(ax, charging_da, discharging_da, alpha_key, title):

    for i in range(24):
        charging = float(charging_da[i])
        discharging = float(discharging_da[i])


        if discharging == 0 and charging != 0:
            state = "charging"
            base_value = charging

        elif discharging != 0 and charging == 0:
            state = "discharging"
            base_value = discharging

        elif discharging == 0 and charging == 0:
            state = "idle"
            base_value = 0.0

        else:
            raise ValueError(
                f"Both DA charging and discharging are nonzero "
                f"at hour {i}, alpha key {alpha_key}."
            )

        ax.bar(
            X[i],
            base_value,
            width,
            color="silver"
        )

        for scenario, offset, color in scenarios:
            ch_plus = (
                scenario_alpha_data_ch_plus[scenario][alpha_key][i]
            )
            ch_minus = (
                scenario_alpha_data_ch_minus[scenario][alpha_key][i]
            )
            dis_plus = (
                scenario_alpha_data_dis_plus[scenario][alpha_key][i]
            )
            dis_minus = (
                scenario_alpha_data_dis_minus[scenario][alpha_key][i]
            )

            if state == "charging":
                arrow_lengths = (
                    ch_plus,
                    -ch_minus + dis_plus
                )

            elif state == "discharging":
                arrow_lengths = (
                    dis_plus,
                    -dis_minus + ch_plus
                )

            else:
                arrow_lengths = (
                    ch_plus,
                    dis_plus
                )

            for dy in arrow_lengths:
                draw_arrow(
                    ax,
                    X[i] + offset,
                    base_value,
                    dy,
                    color
                )

    ax.set_title(title, fontsize=35)
    ax.set_ylabel("Power [MW]", fontsize=35)
    ax.set_ylim(-5, 5)
    ax.set_xticks(X)
    ax.set_xticklabels(X)
    ax.tick_params(
        axis="both",
        which="major",
        labelsize=35
    )
    ax.grid(False)
    ax.grid(True, axis="y")



draw_panel(
    axs[0],
    chDA1,
    disDA1,
    1,
    "α=0"
)

draw_panel(
    axs[1],
    chDA6,
    disDA6,
    6,
    "α=0.5"
)

draw_panel(
    axs[2],
    chDA11,
    disDA11,
    11,
    "α=0.999"
)


axs[2].set_xlabel("Time [h]", fontsize=35)


legend_handles = [
    Patch(facecolor="deeppink", label="S1"),
    Patch(facecolor="darkorange", label="S5"),
    Patch(facecolor="rebeccapurple", label="S10"),
    Patch(
        facecolor="silver",
        label="Charging/discharging in the DA market"
    )
]

axs[0].legend(
    handles=legend_handles,
    loc="upper center",
    bbox_to_anchor=(0, 0.65, 1, 1),
    ncol=4,
    mode="expand",
    fontsize=35
)


fig.savefig(
    "fig_13.pdf",
    format="pdf",
    dpi=100,
    bbox_inches="tight"
)

plt.show()
```
