```python
# Replace with your path if different
df =  pd.read_excel (r'resultsProposed_alpha.xlsx', sheet_name='chBIBminus')

# Initialize nested dictionary
scenario_alpha_data_ch_minus = defaultdict(lambda: defaultdict(list))

#dictionary generation
for i in range(len(df)):
    scenario = df.loc[i, 'scenario']
    alpha = df.loc[i, 'alpha']
    ch_minus = df.loc[i, 'ch_minus']
    scenario_alpha_data_ch_minus[scenario][alpha].append(ch_minus)


# convert to regular dict (for cleaner printing or export)
scenario_alpha_data_ch_minus = {k: dict(v) for k, v in  scenario_alpha_data_ch_minus.items()}

# Print the ch_minus list for scenario 1 and alpha 1
print(scenario_alpha_data_ch_minus[1.0][1.0])
```

```python
# Replace with your path if different
df =  pd.read_excel (r'resultsProposed_alpha.xlsx', sheet_name='chBIBplus')

# Initialize nested dictionary
scenario_alpha_data_ch_plus = defaultdict(lambda: defaultdict(list))

#dictionary generation
for i in range(len(df)):
    scenario = df.loc[i, 'scenario']
    alpha = df.loc[i, 'alpha']
    ch_plus = df.loc[i, 'ch_plus']
    scenario_alpha_data_ch_plus[scenario][alpha].append(ch_plus)


# convert to regular dict (for cleaner printing or export)
scenario_alpha_data_ch_plus = {k: dict(v) for k, v in scenario_alpha_data_ch_plus.items()}

# Print the ch_plus list for scenario 1 and alpha 1
print(scenario_alpha_data_ch_plus[1.0][1.0])
```

```python
# Replace with your path if different
df =  pd.read_excel (r'resultsProposed_alpha.xlsx', sheet_name='disBIBplus')

# Initialize nested dictionary
scenario_alpha_data_dis_plus = defaultdict(lambda: defaultdict(list))

#dictionary generation
for i in range(len(df)):
    scenario = df.loc[i, 'scenario']
    alpha = df.loc[i, 'alpha']
    dis_plus = df.loc[i, 'dis_plus']

    scenario_alpha_data_dis_plus[scenario][alpha].append(-dis_plus)

# convert to regular dict (for cleaner printing or export)
scenario_alpha_data_dis_plus = {k: dict(v) for k, v in  scenario_alpha_data_dis_plus.items()}

# Print the dis_plus list for scenario 1 and alpha 1
print(scenario_alpha_data_dis_plus[1.0][1.0])
```
