```python
import pandas as pd 
from matplotlib import pyplot as plt
import numpy as np
from collections import defaultdict
from matplotlib.patches import Patch
```

```python
#replace with your path of resultsHybridSPRO_Gamma.xlsx file if different
ch_DA_gamma0 = pd.read_excel (r'resultsHybridSPRO_Gamma.xlsx', sheet_name='chBDA')
ch_DA_gamma0 = ch_DA_gamma0['chDA']
ch_DA_gamma0 = ch_DA_gamma0[0:]
dis_DA_gamma0 = pd.read_excel (r'resultsHybridSPRO_Gamma.xlsx', sheet_name='disBDA')
dis_DA_gamma0 = dis_DA_gamma0['disDA']
dis_DA_gamma0 = dis_DA_gamma0[0:]


```

```python
chDA1 = []
chDA12 = []
chDA24 = []


disDA1 = []
disDA12 = []
disDA24 = []
```

```python
for i in range (0, len(ch_DA_gamma0), 25):
    chDA1.append(ch_DA_gamma0[i])
    chDA12.append(ch_DA_gamma0[i+12])
    chDA24.append(ch_DA_gamma0[i+24])

    disDA1.append(-dis_DA_gamma0[i])
    disDA12.append(-dis_DA_gamma0[i+12])
    disDA24.append(-dis_DA_gamma0[i+24])

    
    
```

```python
# Replace with your path if different
df =  pd.read_excel (r'resultsHybridSPRO_Gamma.xlsx', sheet_name='chBIBplus')

# Initialize nested dictionary
scenario_alpha_data_ch_plus = defaultdict(lambda: defaultdict(list))

#dictionary generation
for i in range(len(df)):
    scenario = df.loc[i, 'scenario']
    alpha = df.loc[i, 'alpha']
    ch_plus = df.loc[i, 'ch_plus']
    scenario_alpha_data_ch_plus[scenario][alpha].append(ch_plus)
        

#convert to regular dict (for cleaner printing or export)
scenario_alpha_data_ch_plus = {k: dict(v) for k, v in scenario_alpha_data_ch_plus.items()}

#print the ch_plus list for scenario 1 and alpha 1 
print(scenario_alpha_data_ch_plus[1.0][1.0])
```

```python
# Replace with your path if different
df =  pd.read_excel (r'resultsHybridSPRO_Gamma.xlsx', sheet_name='chBIBminus')

scenario_alpha_data_ch_minus = defaultdict(lambda: defaultdict(list))

for i in range(len(df)):
    scenario = df.loc[i, 'scenario']
    alpha = df.loc[i, 'alpha']
    ch_minus = df.loc[i, 'ch_minus']
    scenario_alpha_data_ch_minus[scenario][alpha].append(ch_minus)


#convert to regular dict (for cleaner printing or export)
scenario_alpha_data_ch_minus = {k: dict(v) for k, v in  scenario_alpha_data_ch_minus.items()}

#print the ch_minus list for scenario 1 and alpha 1
print(scenario_alpha_data_ch_minus[1.0][1.0])
```

```python
# Replace with your path if different
df =  pd.read_excel (r'resultsHybridSPRO_Gamma.xlsx', sheet_name='disBIBplus')

scenario_alpha_data_dis_plus = defaultdict(lambda: defaultdict(list))

for i in range(len(df)):
    scenario = df.loc[i, 'scenario']
    alpha = df.loc[i, 'alpha']
    dis_plus = df.loc[i, 'dis_plus']
    scenario_alpha_data_dis_plus[scenario][alpha].append(-dis_plus)


#convert to regular dict (for cleaner printing or export)
scenario_alpha_data_dis_plus = {k: dict(v) for k, v in  scenario_alpha_data_dis_plus.items()}

#print the dis_plus list for scenario 1 and alpha 1
print(scenario_alpha_data_dis_plus[1.0][1.0])
```

```python
# Replace with your path if different
df =  pd.read_excel (r'resultsHybridSPRO_Gamma.xlsx', sheet_name='disBIBminus')


scenario_alpha_data_dis_minus = defaultdict(lambda: defaultdict(list))

for i in range(len(df)):
    scenario = df.loc[i, 'scenario']
    alpha = df.loc[i, 'alpha']
    dis_minus = df.loc[i, 'dis_minus']
    if (dis_minus>0.15):
         scenario_alpha_data_dis_minus[scenario][alpha].append(-dis_minus)
    else: 
        scenario_alpha_data_dis_minus[scenario][alpha].append(0)

#convert to regular dict (for cleaner printing or export)
scenario_alpha_data_dis_minus = {k: dict(v) for k, v in  scenario_alpha_data_dis_minus.items()}

#print the dis_minus list for scenario 1 and alpha 1
print(scenario_alpha_data_dis_minus[1.0][1.0])
```

```python
X = np.arange(0, 25)


width = 0.9
hw = 0.2
hl = 0.3
arrow_width = 0.05
zero_tolerance = 1e-9

fig, axs = plt.subplots(
    3,
    1,
    figsize=(25, 15),
    constrained_layout=True
)


scenarios = [
    (1, -0.33, "deeppink"),
    (5,  0.00, "darkorange"),
    (10, 0.33, "rebeccapurple")
]


def is_zero(value):
    return np.isclose(
        float(value),
        0.0,
        atol=zero_tolerance,
        rtol=0.0
    )


def draw_arrow(ax, x, y, dy, color):
    dy = float(dy)

    if not is_zero(dy):
        ax.arrow(
            x,
            y,
            0,
            dy,
            color=color,
            length_includes_head=True,
            head_width=hw,
            head_length=hl,
            width=arrow_width
        )


def draw_panel(
    ax,
    charging_da,
    discharging_da,
    gamma_position,
    title
):
   

    for i in range(24):

        charging_value = float(charging_da[i])
        discharging_value = float(discharging_da[i])

        charging_is_zero = is_zero(charging_value)
        discharging_is_zero = is_zero(discharging_value)

        # Charging in the day-ahead market
        if discharging_is_zero and not charging_is_zero:

            base_value = charging_value

            ax.bar(
                X[i],
                base_value,
                width,
                color="silver"
            )

            for scenario, offset, color in scenarios:

                first_arrow = (
                    scenario_alpha_data_ch_plus
                    [scenario][gamma_position][i]
                )

                second_arrow = (
                    -scenario_alpha_data_ch_minus
                    [scenario][gamma_position][i]
                    +
                    scenario_alpha_data_dis_plus
                    [scenario][gamma_position][i]
                )

                draw_arrow(
                    ax,
                    X[i] + offset,
                    base_value,
                    first_arrow,
                    color
                )

                draw_arrow(
                    ax,
                    X[i] + offset,
                    base_value,
                    second_arrow,
                    color
                )

        # Discharging in the day-ahead market
        elif not discharging_is_zero and charging_is_zero:

            base_value = discharging_value

            ax.bar(
                X[i],
                base_value,
                width,
                color="silver"
            )

            for scenario, offset, color in scenarios:

                first_arrow = (
                    scenario_alpha_data_dis_plus
                    [scenario][gamma_position][i]
                )

                second_arrow = (
                    -scenario_alpha_data_dis_minus
                    [scenario][gamma_position][i]
                    +
                    scenario_alpha_data_ch_plus
                    [scenario][gamma_position][i]
                )

                draw_arrow(
                    ax,
                    X[i] + offset,
                    base_value,
                    first_arrow,
                    color
                )

                draw_arrow(
                    ax,
                    X[i] + offset,
                    base_value,
                    second_arrow,
                    color
                )

        # No charging or discharging in the day-ahead market
        elif charging_is_zero and discharging_is_zero:

            base_value = 0.0

            ax.bar(
                X[i],
                base_value,
                width,
                color="silver"
            )

            for scenario, offset, color in scenarios:

                charging_arrow = (
                    scenario_alpha_data_ch_plus
                    [scenario][gamma_position][i]
                )

                discharging_arrow = (
                    scenario_alpha_data_dis_plus
                    [scenario][gamma_position][i]
                )

                draw_arrow(
                    ax,
                    X[i] + offset,
                    base_value,
                    charging_arrow,
                    color
                )

                draw_arrow(
                    ax,
                    X[i] + offset,
                    base_value,
                    discharging_arrow,
                    color
                )


        else:
            raise ValueError(
                f"Both charging and discharging are nonzero at hour {i}: "
                f"charging={charging_value}, "
                f"discharging={discharging_value}"
            )

    
    ax.set_title(title, fontsize=35)
    ax.set_ylabel("Power [MW]", fontsize=35)
    ax.set_ylim(-5, 5)
    ax.set_xticks(np.arange(len(X)))
    ax.set_xticklabels(X)
    ax.tick_params(
        axis="both",
        which="major",
        labelsize=35
    )
    ax.grid(True, axis="y")



draw_panel(
    axs[0],
    chDA1,
    disDA1,
    1,
    r"$\Gamma^{IB}$=0"
)

draw_panel(
    axs[1],
    chDA12,
    disDA12,
    13,
    r"$\Gamma^{IB}$=12"
)

draw_panel(
    axs[2],
    chDA24,
    disDA24,
    25,
    r"$\Gamma^{IB}$=24"
)

axs[2].set_xlabel("Time [h]", fontsize=35)


legend_handles = [
    Patch(facecolor="deeppink", label="S1"),
    Patch(facecolor="darkorange", label="S5"),
    Patch(facecolor="rebeccapurple", label="S10"),
    Patch(
        facecolor="silver",
        label="Charging/discharging in the DA market"
    )
]

axs[0].legend(
    handles=legend_handles,
    loc="upper center",
    bbox_to_anchor=(0, 0.65, 1, 1),
    ncol=4,
    mode="expand",
    fontsize=35
)


fig.savefig(
    "fig_12.pdf",
    format="pdf",
    dpi=100,
    bbox_inches="tight"
)

plt.show()
```
